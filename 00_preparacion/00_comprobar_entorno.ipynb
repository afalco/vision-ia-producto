```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

imagen = np.zeros((128, 128, 3), dtype=np.uint8)
imagen[24:104, 32:96] = [40, 140, 220]
assert np.asarray(Image.fromarray(imagen)).shape == (128, 128, 3)
plt.imshow(imagen)
plt.title("Pieza sintética: entorno preparado")
plt.axis("off")
plt.show()
```
